# 🗺️ CURSUS DE PROFESSIONNALISATION PYTHON

## 🪓 MODULE 1 : **Paramètres Avancés des Fonctions & Décorateurs**

**🎯 Objectif :** *Maîtriser la manipulation dynamique des arguments et la modification de comportement à la volée.*

* **Arguments positionnels vs nommés :** Syntaxe, ordre d'appel et contraintes.
* **Dépaquetage `*args` et `**kwargs` :** Fonctions à paramètres infinis et ouverture de collections.
* **Paramètres restrictifs :** Usage des symboles `/` (positional-only) et `*` (keyword-only).
* **Fondations fonctionnelles :** Fonctions de première classe, imbrication et *closures*.
* **Sucre syntaxique `@` :** Mécanique d'application des décorateurs.
* **Décorateurs universels :** Combinaison avec `*args` / `kwargs` (Ex: création d'un `@chronometre`).

---

## 🏛️ MODULE 2 : **POO Avancée & Abstraction** (Propriétés & Classes Abstraites)

**🎯 Objectif :** *Sécuriser l'encapsulation et imposer des contrats d'interface stricts.*

* **Encapsulation pythonique :** Remplacement des getters/setters par le décorateur `@property`.
* **Validation de données :** Interception des modifications via `@attribut.setter`.
* **Module `abc` :** Utilisation de la classe `ABC` et du décorateur `@abstractmethod`.
* **Classes Abstraites :** Interdiction d'instanciation de la classe mère et obligation d'implémentation dans les classes filles.

---

## 📐 MODULE 3 : **Qualité du Code, Standardisation & Documentation**

**🎯 Objectif :** *Produire un code lisible, typé et auto-documenté selon les normes industrielles.*

* **Notation des types (Type Hinting) :** Annotations de variables et signatures de fonctions.
* **Module `typing` :** Usage de `List`, `Dict`, `Optional` et `Union`.
* **Docstrings normalisées :** Rédaction aux formats *Google Style* ou *Sphinx* (`Args:`, `Returns:`, `Raises:`).
* **Conventions PEP 8 :** Règles de nommage (snake_case, PascalCase) et règles d'espacement.

---

## ⚡ MODULE 4 : **Le Code "Pythonique"** (Raccourcis & Optimisation)

**🎯 Objectif :** *Écrire un code concis, performant et élégant.*

* **Compréhensions :** Filtrage et transformation de listes et dictionnaires en une seule ligne.
* **Opérateur ternaire :** Simplification des structures `if/else` d'assignation.
* **Fonctions anonymes :** Syntaxe `lambda` et tris complexes basés sur des clés (`.sort(key=...)`).

---

## 📦 MODULE 5 : **L'Écosystème Moderne & Gestion des Dépendances**

**🎯 Objectif :** *Isoler les projets et gérer proprement les bibliothèques externes.*

* **Environnements virtuels (`venv`) :** Création, activation, désactivation.
* **Gestionnaire `pip` :** Installation de paquets via PyPI et gestion des versions.
* **Fichier `requirements.txt` :** Figer (`pip freeze`) et installer (`pip install -r`) les dépendances.

---

## 🏢 MODULE 6 : **Persistance Industrielle & Modélisation Relationnelle** (SQLite3)

**🎯 Objectif :** *Passer des fichiers plats (JSON) à une base de données relationnelle structurée.*

* **Modélisation SQL :** Tables, types de données (`TEXT`, `INTEGER`, `REAL`).
* **Clé Primaire (`PRIMARY KEY`) :** Unicité des enregistrements et auto-incrémentation.
* **Clés Étrangères (`FOREIGN KEY`) :** Mise en place de relations 1:N entre les tables.
* **Intégrité des données :** Contraintes `NOT NULL`, `UNIQUE` et `CHECK`.
* **Requêtes CRUD :** Syntaxes `INSERT INTO`, `SELECT` (+ `WHERE`, `LIKE`, `ORDER BY`), `UPDATE` et `DELETE`.
* **Connecteur `sqlite3` :** Connexion, Curseur (`execute`, `fetchall`), `commit()` et `rollback()`.

---

**🎯 Objectif :** *Développer une application bureau cliquable moderne.*

* **CustomTkinter :** Fenêtres, responsive (layout grid), boutons, champs de saisie (`CTkEntry`) et menus déroulants.
  - intégration des thèmes (mode clair-mode sombre)

## 📦 MODULE 8 : **Déploiement & Distribution Industrielle**

**🎯 Objectif :** *Packager l'application pour une distribution client finale sans prérequis.*

* **Compilation PyInstaller :** Génération d'un exécutable unique via la commande `pyinstaller --onefile --noconsole main.py`.
* **Recette de déploiement :** Validation et exécution du fichier `.exe` sur une machine vierge sans Python.
