## 1. **NOTATION DES TYPES** (*Type Hinting*)

En Python, les types sont dynamiques : on n'a pas besoin de déclarer si une variable est une chaîne (`str`) ou un entier (`int`). Cependant, dans des projets d'entreprise, ajouter des **annotations de type** apporte trois gros avantages :

* **Clarté immédiate :** Vos collègues (et vous-même dans 6 mois) savent exactement quelles données sont attendues.
* **Autocomplétion dans l'IDE :** VS Code ou PyCharm vous proposent les bonnes méthodes automatiques (ex: méthodes de `str` si la variable est typée `: str`).
* **Détection des bugs avant l'exécution :** Des outils comme `mypy` peuvent analyser votre code et vous avertir des erreurs de type.

---

#### Syntaxe de base

**Variables simples :**

```python
nom: str = "Hulrich"
age: int = 25
solde: float = 1250.50
est_actif: bool = True

```

**Signatures de fonctions :**
On précise le type de chaque paramètre après un deux-points `:`, et le type de retour après une flèche `->`.

```python
def calculer_remise(prix_initial: float, pourcentage: float) -> float:
    return prix_initial * (1 - pourcentage)

```

Si une fonction ne renvoie rien (elle fait seulement un `print` ou modifie un état), on annotera son retour avec `-> None` :

```python
def afficher_message(message: str) -> None:
    print(f"📣 {message}")

```

## 2. **ANNOTATION DES COLLECTIONS & TYPES AVANCÉS**

Pour annoter des structures de données complexes (listes, dictionnaires, structures optionnelles), Python permet d'utiliser les types natifs, l'opérateur d'union `|`, ainsi que les abstractions spécialisées du module `typing`.

### a. **Les Conteneurs** (Listes, Dictionnaires, Tuples)

On précise le type des éléments contenus entre crochets `[]` directement sur le type natif :

```python
# Une liste contenant uniquement des entiers
nombres: list[int] = [10, 20, 30]

# Un dictionnaire homogène (toutes les valeurs sont du même type)
catalogue: dict[str, float] = {"Pomme": 1.5, "Banane": 0.8}

# Un tuple à structure fixe (nom, âge, note)
etudiant: tuple[str, int, float] = ("Hulrich", 25, 17.5)

```

### b. **Les Dictionnaires Structurés** (`TypedDict`)

Quand un dictionnaire représente un objet avec **des clés fixes ayant des types différents**, `dict[str, Any]` n'est pas assez précis. On importe `TypedDict` depuis le module `typing` pour créer un vrai schéma de dictionnaire.

```python
from typing import TypedDict


# Définition de la structure du dictionnaire
class Article(TypedDict):
  nom: str
  prix: float
  en_stock: bool


# Utilisation du TypedDict
produit: Article = {"nom": "Clavier", "prix": 45.0, "en_stock": True}


# Typage d'une liste de dictionnaires structurés
def calculer_panier(items: list[Article]) -> float:
  return sum(item["prix"] for item in items if item["en_stock"])

```

> 💡 **Pourquoi `TypedDict` ?**
> * L'IDE active l'autocomplétion sur les clés (`item["nom"]`, `item["prix"]`).
>  
> 


### c. **Valeurs Multiples & Valeurs Optionnelles** (`|`)

Lorsque le paramètre ou le retour d'une fonction peut accepter plusieurs types, on utilise l'opérateur `|` :

* **Plusieurs types possibles :** `int | float` (accepte un entier ou un nombre à virgule).
* **Valeur optionnelle :** `str | None` (indique que la valeur peut être une chaîne de caractères ou absente).

```python
def chercher_produit(reference: int | str) -> Article | None:
  """Recherche un produit par son ID (int) ou sa référence (str).

  Renvoie le dictionnaire du produit typé ou None si introuvable.
  """
  if reference == 101 or reference == "REF-101":
    return {"nom": "Souris", "prix": 25.0, "en_stock": True}
  return None

```


### d. **Typage des Fonctions transmises en argument** (`Callable`)

Quand une fonction reçoit une autre fonction en paramètre (comme dans un décorateur ou un callback), on importe `Callable` du module `typing` :

```python
from typing import Callable


def executer_action(
    valeur: float, operation: Callable[[float], float]
) -> float:
  # operation attend un float et renvoie un float
  return operation(valeur)

```

## 3. **DOCSTRINGS NORMALISÉES** (Format Google Style)

### 📌 **Qu'est-ce qu'une Docstring** ?

Une **docstring** (chaîne de documentation) est un bloc de texte placé **immédiatement sous la signature** d'une fonction, d'une classe ou d'un module.

Contrairement aux commentaires classiques commençant par `#`, la docstring est lue par Python à l'exécution : elle est accessible via la fonction `help(ma_fonction)` ou l'attribut `ma_fonction.__doc__`, et elle apparaît directement dans l'autocomplétion des IDE (VS Code, PyCharm).

---

### 🏛️ **La Syntaxe & Structure Officielle** (Google Style)

En entreprise, le format **Google Style** est le standard le plus répandu pour sa lisibilité. Il s'écrit entre trois guillemets double `"""` et s'organise toujours selon les **4 blocs fondamentaux** ci-dessous :

```python
def nom_de_la_fonction(param1: type1, param2: type2 = valeur_par_defaut) -> type_retour:
    """Résumé en une ligne résumant l'action de la fonction (au présent de l'indicatif).

    Description plus détaillée si nécessaire. On explique ici le comportement
    global, le contexte d'utilisation ou la logique métier sous-jacente.

    Args:
        param1 (type1): Explication du premier paramètre et ses contraintes.
        param2 (type2, optional): Explication du second paramètre. 
            Préciser la valeur par défaut si elle existe (ex: Defaults to 10).

    Returns:
        type_retour: Description de ce que la fonction renvoie concrètement.

    Raises:
        NomDeLException: Condition sous laquelle cette exception est levée.
    """
    # Corps de la fonction...

```


### 🔍 **Décortiquage des 4 Blocs** :

| Bloc | Obligatoire ? | Rôle & Règles de rédaction |
| --- | --- | --- |
| **1. Résumé (Ligne 1)** | **Oui** | Une phrase courte au présent de l'indicatif (ex: *Calcule le total...*, *Convertit les données...*). Elle se termine par un point. |
| **2. `Args`:** | **Oui** *(si paramètres)* | Indique le nom de chaque paramètre, son **type entre parenthèses**, et son rôle. Si le paramètre a une valeur par défaut, on ajoute `optional` et `Defaults to X`. |
| **3. `Returns`:** | **Oui** *(si retour $\neq$ None)* | Indique le **type du retour**, suivi de deux-points `:` et de la description de la valeur renvoyée. |
| **4. `Raises`:** | Non *(seulement si `raise`)* | Liste les exceptions explicites soulevées par la fonction en cas d'erreur de saisie ou d'exécution. |


### 💻 **Exemple Modèle Complet**

Voici un modèle d'illustration concret d'une fonction documentée avec cette syntaxe :

```python
def diviser_capital(montant_total: float, nombre_parts: int) -> float:
    """Divise un montant global en parts égales entre plusieurs bénéficiaires.

    Args:
        montant_total (float): La somme totale à répartir en dollars.
        nombre_parts (int): Le nombre de bénéficiaires (doit être > 0).

    Returns:
        float: La valeur exacte de la part attribuée à chaque bénéficiaire.

    Raises:
        ValueError: Si le montant_total est négatif.
        ZeroDivisionError: Si nombre_parts est égal à zéro.
    """
    if montant_total < 0:
        raise ValueError("Le montant total ne peut pas être négatif.")
    
    return montant_total / nombre_parts

```

## 5. **NORMES PEP 8 & BONNES PRATIQUES DE STYLE**

### **Qu'est-ce que la PEP 8 ?**

La **PEP 8** (*Python Enhancement Proposal #8*) est le guide de style officiel pour le code Python. Son objectif principal est de rendre le code **lisible, homogène et maintenable**, quel que soit le développeur qui le lit ou l'écrit.


### **Les 5 Règles Fondamentales de la PEP 8**

```
Normes PEP 8
 ├── 1. Conventions de nommage (Naming)
 ├── 2. Structure et espacement (Formatting)
 ├── 3. Organisation des imports (Imports Order)
 ├── 4. Espacement autour des opérateurs (Whitespace)
 └── 5. Longueur des lignes (Line Length)

```

#### 1. **Conventions de Nommage**

Python exige une casse spécifique selon la nature de l'élément :

| Élément | Style de nommage | Exemple conforme |
| --- | --- | --- |
| **Variables & Fonctions** | `snake_case` (minuscules + underscores) | `solde_compte`, `calculer_total()` |
| **Classes** | `PascalCase` (majuscule à chaque mot) | `CompteBancaire`, `GestionnairePaie` |
| **Constantes** | `UPPER_SNAKE_CASE` (MAJUSCULES + underscores) | `TAUX_TVA_PAR_DEFAUT = 0.16` |
| **Modules / Fichiers** | `snake_case` (court, sans majuscule) | `traitement_donnees.py` |
| **Attributs privés** | Préfixe par un underscore `_` | `_solde_interne` |


#### 2. **Structure et Lignes Vides**

Pour maintenir une mise en page aérée et lisible :

* **Indentation :** Toujours **4 espaces par niveau** (jamais de tabulations).
* **2 lignes vides :** Autour des fonctions globales et des définitions de classes.
* **1 ligne vide :** Entre les méthodes au sein d'une même classe.

```python
# 2 lignes vides avant la classe
class AnalyseurDonnees:

  def __init__(self, nom: str) -> None:
    self.nom = nom

  # 1 ligne vide entre deux méthodes
  def valider(self) -> bool:
    return len(self.nom) > 0


# 2 lignes vides entre la classe et la fonction globale
def executer_analyse() -> None:
  pass

```

#### 3. **Organisation des Imports**

Tous les `import` se placent en haut du fichier, découpés en **3 blocs distincts** séparés par une ligne vide :

1. **Bibliothèque standard Python** (`os`, `math`, `sys`, `typing`, etc.)
2. **Bibliothèques tierces installées** (`pandas`, `numpy`, `requests`, etc.)
3. **Modules locaux du projet** (`from services.paie import calculer_salaire`)

```python
# 1. Bibliothèque standard
import os
from typing import Callable

# 2. Bibliothèques tierces
import numpy as np
import pandas as pd

# 3. Modules locaux
from services.donnees import charger_fichier

```

#### 4. **Espacement autour des Opérateurs & Annotations**

* **Opérateurs d'affectation et de comparaison :** Un espace de chaque côté (`=`, `==`, `+`, `-`, `*`, `/`).
* **Valeurs par défaut dans les paramètres :** Pas d'espace autour du `=` (`taux: float = 0.16`).
* **Annotations de type :** Pas d'espace avant le deux-points `:`, un espace après (`nom: str`).

```python
# ❌ INCORRECT (Non conforme PEP 8)
def calculer(prix,taux=0.16):
  total = prix*(1+taux)
  return total


# ✅ CORRECT (Conforme PEP 8)
def calculer(prix: float, taux: float = 0.16) -> float:
  total = prix * (1 + taux)
  return total

```

#### 5. **Longueur Maximale des Lignes**

La PEP 8 fixe une limite de **79 caractères** par ligne (portée à 88 ou 100 dans les projets utilisant des formatteurs automatisés modernes comme `Black`).
