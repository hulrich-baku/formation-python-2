## 5.1 <u>**MODULES, IMPORTS ET LE GESTIONNAIRE `pip`**</u>

### 1. **Rappel : Importer ses Propres Fichiers (`.py`)**

En Python, **chaque fichier `.py` est un module**. Pour éviter d'avoir un fichier de 1000 lignes, on découpe son code en plusieurs fichiers et on les réutilise grâce au mot-clé `import`.

#### Exemple Concret :

Imaginons la structure de projet suivante :

```text
mon_projet/
├── outils.py
└── main.py

```

#### **Le fichier `outils.py` (contient nos fonctions) :**

```python
def generer_pseudo(prenom: str, age: int) -> str:
  """Génère un pseudonyme pour un joueur."""
  return f"{prenom.lower()}_{age}"
```


#### **Le fichier `main.py` (le programme principal) :**

Il existe **3 façons** d'importer et d'utiliser ce module :

```python
# Option A : Importer tout le module
import outils

pseudo = outils.generer_pseudo("Hulrich", 25)

# Option B : Importer une fonction précise (Recommandé)
from outils import generer_pseudo

pseudo = generer_pseudo("Hulrich", 25)

# Option C : Importer avec un alias (un surnom)
import outils as out

pseudo = out.generer_pseudo("Hulrich", 25)
```

## Step 1 : **Colorama** (Couleurs et Styles dans le Terminal)

Par défaut, la console écrit en blanc sur fond noir. **Colorama** permet d'ajouter très simplement des couleurs au texte, au fond (*background*), ainsi que des styles (texte en gras, atténué, etc.).

### 1. **Installation**

Dans le terminal :

```bash
pip install colorama

```

### 2. **Les 3 Composants Principaux**

* **`Fore`** *(Foreground)* : Change la couleur du **texte** (`RED`, `GREEN`, `YELLOW`, `BLUE`, `CYAN`, `MAGENTA`, `WHITE`, `RESET`).
* **`Back`** *(Background)* : Change la couleur de **fond** du texte (`Back.RED`, `Back.YELLOW`, `RESET`, etc.).
* **`Style`** : Modifie le style du texte (`BRIGHT` pour le gras/brillant, `RESET_ALL` pour tout remettre par défaut).

Voici les différents attributs que peuvent avoir la couleur de texte (`Fore`) ou la couleur de fond (`Back`) :

| Valeur   | Explication                                                                 |
|----------|-----------------------------------------------------------------------------|
| BLACK    | Texte/Fond en noir.                                                         |
| RED      | Texte/Fond en rouge (erreurs, alertes).                                     |
| GREEN    | Texte/Fond en vert (succès, validations).                                   |
| YELLOW   | Texte/Fond en jaune (avertissements, infos importantes).                    |
| BLUE     | Texte/Fond en bleu (infos, messages neutres).                               |
| MAGENTA  | Texte/Fond en magenta (accent, messages spéciaux).                          |
| CYAN     | Texte/Fond en cyan (infos techniques, logs).                                |
| WHITE    | Texte/Fond en blanc (texte clair sur fond sombre).                          |
| RESET    | Réinitialise la couleur du Texte/Fond à la valeur par défaut du terminal.   |

Les différentes valeurs que peut avoir le style (`Style`) de texte :

| Valeur   | Explication                                                       |
|----------|-------------------------------------------------------------------|
| DIM      | Texte atténué (moins intense, plus discret).                      |
| NORMAL   | Texte en style normal (annule `DIM` ou `BRIGHT`).                 |
| BRIGHT   | Texte intense / plus lumineux (plus visible).                     |
| RESET_ALL| Réinitialise tout : couleurs (Fore/Back) et style (DIM/BRIGHT).   |


### 3. **La Réinitialisation Automatique** (Règle d'or)

Si vous oubliez d'annuler une couleur, tout le texte qui suit dans le terminal restera coloré !
Pour éviter cela, on utilise `init(autoreset=True)` dès le début du programme.

```python
from colorama import Back, Fore, Style, init

# Réinitialise la couleur automatiquement après chaque print()
init(autoreset=True)

# Texte vert
print(Fore.GREEN + "✅ Succès : Le fichier a été sauvegardé avec succès !")

# Texte rouge avec fond jaune
print(
    Back.YELLOW
    + Fore.RED
    + "⚠️ Avertissement : Le fichier est bientôt plein."
)

# Texte brillant/gras
print(Style.BRIGHT + Fore.CYAN + "ℹ️ Information : Connexion établie.")

```

## 🎭 Step 2 : **Faker** (Génération de Données Factices)

Lorsque l'on développe une application (système d'école, gestion de stocks, formulaires), on a continuellement besoin de jeux de données pour tester nos boucles, filtres et fonctions.

Saisir manuellement des noms et adresses prend du temps. **Faker** résout ce problème en générant instantanément des noms, adresses, emails, numéros de téléphone et textes réalistes.


### 1. **Installation**

Dans le terminal :

```bash
pip install faker
```


### 2. **Prise en main de Faker**

#### a. **Utilisation de base**

```py
from faker import Faker

# Instanciation de l'objet Faker
fake = Faker()

print(fake.name())          # ex: "John Doe"
print(fake.email())         # ex: "jdoe@example.com"
print(fake.address())       # adresse complète
print(fake.phone_number())  # numéro de téléphone
print(fake.job())           # Métier
```
Par défaut, les données générées sont en Anglais, la langue par défaut de python.

#### b. **Personalisation selon la région**

Pour générer des données adaptées à une région spécifique, on initialise `Faker` en précisant le code de langue (ex: `"fr_FR"` pour le français).

```python
from faker import Faker

# Initialisation du générateur en français
fake = Faker("fr_FR")

# Génération de données simples
print(f"Nom complet : {fake.name()}")
print(f"Adresse     : {fake.address()}")
print(f"Email       : {fake.email()}")
print(f"Téléphone   : {fake.phone_number()}")
print(f"Metier      : {fake.job()}")

```

### 3. **Générer des Listes de Données Structurées**

Combiné aux **compréhensions de listes** et aux **`TypedDict`** (vus aux modules précédents), Faker permet de fabriquer de véritables bases de données de test en 3 lignes de code !

```python
from typing import TypedDict
from faker import Faker

fake = Faker("fr_FR")

class Etudiant(TypedDict):
    nom: str
    email: str
    note: int

# Génération d'une liste de 5 étudiants factices
promotion: list[Etudiant] = [
    {
        "nom": fake.name(),
        "email": fake.email(),
        "note": fake.random_int(min=0, max=20)
    }
    for _ in range(5)
]

for etudiant in promotion:
    print(etudiant)

```
