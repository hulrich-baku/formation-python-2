## 5.1 <u>**MODULES, IMPORTS ET LE GESTIONNAIRE `pip`**</u>

### 1. **Rappel : Importer ses Propres Fichiers (`.py`)**

En Python, **chaque fichier `.py` est un module**. Pour éviter d'avoir un fichier de 1000 lignes, on découpe son code en plusieurs fichiers et on les réutilise grâce au mot-clé `import`.

#### Exemple Concret :

Imaginons la structure de projet suivante :

```text
mon_projet/
├── outils.py
└── main.py

```

#### **Le fichier `outils.py` (contient nos fonctions) :**

```python
def generer_pseudo(prenom: str, age: int) -> str:
  """Génère un pseudonyme pour un joueur."""
  return f"{prenom.lower()}_{age}"
```


#### **Le fichier `main.py` (le programme principal) :**

Il existe **3 façons** d'importer et d'utiliser ce module :

```python
# Option A : Importer tout le module
import outils

pseudo = outils.generer_pseudo("Hulrich", 25)

# Option B : Importer une fonction précise (Recommandé)
from outils import generer_pseudo

pseudo = generer_pseudo("Hulrich", 25)

# Option C : Importer avec un alias (un surnom)
import outils as out

pseudo = out.generer_pseudo("Hulrich", 25)
```

Quand la bibliothèque standard ne suffit pas, on télécharge des paquets créés par la communauté sur **[PyPI (Python Package Index)](https://pypi.org/)**.

Pour cela, on utilise le terminal et l'outil **`pip`** (le gestionnaire de paquets officiel de Python).

#### **Les Commandes Essentielles de `pip` dans le Terminal** :

| Action | Commande Terminal | Description / Exemple |
| --- | --- | --- |
| **Installer un paquet** | `pip install bilbiotheque` | Télécharge et installe le paquet (ex: `colorama` pour afficher du texte en couleur dans le terminal. On doit executer la commande `pip install colorama`) |

| **Désinstaller** | `pip uninstall bibliotheque` | Supprime le paquet de la machine. (exempel : `pip uninstall colorama`) |

## Step 1 : **Colorama** (Couleurs et Styles dans le Terminal)

Par défaut, la console écrit en blanc sur fond noir. **Colorama** permet d'ajouter très simplement des couleurs au texte, au fond (*background*), ainsi que des styles (texte en gras, atténué, etc.).

### 1. **Installation**

Dans le terminal :

```bash
pip install colorama

```

### 2. **Les 3 Composants Principaux**

* **`Fore`** *(Foreground)* : Change la couleur du **texte** (`RED`, `GREEN`, `YELLOW`, `BLUE`, `CYAN`, `MAGENTA`, `WHITE`, `RESET`).
* **`Back`** *(Background)* : Change la couleur de **fond** du texte (`Back.RED`, `Back.YELLOW`, `RESET`, etc.).
* **`Style`** : Modifie le style du texte (`BRIGHT` pour le gras/brillant, `RESET_ALL` pour tout remettre par défaut).

Voici les différents attributs que peuvent avoir la couleur de texte (`Fore`) ou la couleur de fond (`Back`) :

| Valeur   | Explication                                                                 |
|----------|-----------------------------------------------------------------------------|
| BLACK    | Texte/Fond en noir.                                                         |
| RED      | Texte/Fond en rouge (erreurs, alertes).                                     |
| GREEN    | Texte/Fond en vert (succès, validations).                                   |
| YELLOW   | Texte/Fond en jaune (avertissements, infos importantes).                    |
| BLUE     | Texte/Fond en bleu (infos, messages neutres).                               |
| MAGENTA  | Texte/Fond en magenta (accent, messages spéciaux).                          |
| CYAN     | Texte/Fond en cyan (infos techniques, logs).                                |
| WHITE    | Texte/Fond en blanc (texte clair sur fond sombre).                          |
| RESET    | Réinitialise la couleur du Texte/Fond à la valeur par défaut du terminal.   |

Les différentes valeurs que peut avoir le style (`Style`) de texte :

| Valeur   | Explication                                                       |
|----------|-------------------------------------------------------------------|
| DIM      | Texte atténué (moins intense, plus discret).                      |
| NORMAL   | Texte en style normal (annule `DIM` ou `BRIGHT`).                 |
| BRIGHT   | Texte intense / plus lumineux (plus visible).                     |
| RESET_ALL| Réinitialise tout : couleurs (Fore/Back) et style (DIM/BRIGHT).   |


### 3. **La Réinitialisation Automatique** (Règle d'or)

Si vous oubliez d'annuler une couleur, tout le texte qui suit dans le terminal restera coloré !
Pour éviter cela, on utilise `init(autoreset=True)` dès le début du programme.


```python
from colorama import Back, Fore, Style, init

# Réinitialise la couleur automatiquement après chaque print()
init(autoreset=True)

# Texte vert
print(Fore.GREEN + "✅ Succès : Le fichier a été sauvegardé avec succès !")

# Texte rouge avec fond jaune
print(
    Back.YELLOW
    + Fore.RED
    + "⚠️ Avertissement : Le fichier est bientôt plein."
)

# Texte brillant/gras
print(Style.BRIGHT + Fore.CYAN + "ℹ️ Information : Connexion établie.")

```

## 🎭 Step 2 : **Faker** (Génération de Données Factices)

Lorsque l'on développe une application (système d'école, gestion de stocks, formulaires), on a continuellement besoin de jeux de données pour tester nos boucles, filtres et fonctions.

Saisir manuellement des noms et adresses prend du temps. **Faker** résout ce problème en générant instantanément des noms, adresses, emails, numéros de téléphone et textes réalistes.


### 1. **Installation**

Dans le terminal :


```bash
pip install faker
```


### 2. **Prise en main de Faker**

#### a. **Utilisation de base**


```py
from faker import Faker

# Instanciation de l'objet Faker
fake = Faker()

print(fake.name())          # ex: "John Doe"
print(fake.email())         # ex: "jdoe@example.com"
print(fake.address())       # adresse complète
print(fake.phone_number())  # numéro de téléphone
print(fake.job())           # Métier
```
Par défaut, les données générées sont en Anglais, la langue par défaut de python.

#### b. **Personalisation selon la région**

Pour générer des données adaptées à une région spécifique, on initialise `Faker` en précisant le code de langue (ex: `"fr_FR"` pour le français).


```python
from faker import Faker

# Initialisation du générateur en français
fake = Faker("fr_FR")

# Génération de données simples
print(f"Nom complet : {fake.name()}")
print(f"Adresse     : {fake.address()}")
print(f"Email       : {fake.email()}")
print(f"Téléphone   : {fake.phone_number()}")
print(f"Metier      : {fake.job()}")

```

### 3. **Générer des Listes de Données Structurées**

Combiné aux **compréhensions de listes** et aux **`TypedDict`** (vus aux modules précédents), Faker permet de fabriquer de véritables bases de données de test en 3 lignes de code !


```python
from typing import TypedDict
from faker import Faker

fake = Faker("fr_FR")

class Etudiant(TypedDict):
    nom: str
    email: str
    note: int

# Génération d'une liste de 5 étudiants factices
promotion: list[Etudiant] = [
    {
        "nom": fake.name(),
        "email": fake.email(),
        "note": fake.random_int(min=0, max=20)
    }
    for _ in range(5)
]

for etudiant in promotion:
    print(etudiant)

```

## 🖼️ Step 3 : **Pillow** (Manipulation & Traitement d'Images)

### 1. **Installation**

Dans le terminal :

```bash
pip install pillow

```


### 2. **Les Opérations de Base**

On importe principalement le module `Image` depuis `PIL`.

#### A. **Ouvrir et inspecter une image**

```python
from PIL import Image

# Chargement de l'image
image = Image.open("photo.jpg")

# Inspection des propriétés
print(f"Dimensions : {image.size}")  # Largeur, Hauteur en pixels (ex: (1920, 1080))
print(f"Format     : {image.format}")  # ex: JPEG, PNG
print(f"Mode       : {image.mode}")  # ex: RGB, RGBA

```

#### B. **Redimensionner et Pivoter**

```python
# Redimensionner à 800x600 pixels
image_petite = image.resize((800, 600))

# Pivoter à 90 degrés
image_tournee = image.rotate(90)

# Sauvegarder sous un nouveau nom ou format
image_petite.save("photo_800x600.png")

```

#### C. **Convertir en Noir & Blanc et appliquer des filtres**

```python
from PIL import ImageFilter

# Conversion en niveaux de gris (Noir & Blanc)
image_nb = image.convert("L")

# Application d'un flou
image_floue = image.filter(ImageFilter.BLUR)

image_nb.save("photo_noir_et_blanc.jpg")

```


### 3. **Créer des Images et Dessiner** (`ImageDraw`)

Pillow permet aussi de créer des images vierges de toutes pièces et de dessiner dessus (formes, texte).

```python
from PIL import Image, ImageDraw

# Créer une image de 400x200 pixels avec un fond bleu RGB(0, 102, 204)
image_vierge = Image.new("RGB", (400, 200), color=(0, 102, 204))

# Préparer l'outil de dessin
dessin = ImageDraw.Draw(image_vierge)

# Dessiner un rectangle blanc
dessin.rectangle([20, 20, 380, 180], outline="white", width=3)

# Sauvegarder la création
image_vierge.save("carte_bleue.png")

```

## 5.2 **<u>LE FICHIER `requirements.txt`</u>**

### **Pourquoi ce fichier est indispensable ?**

Maintenant que tu sais installer des bibliothèques externes avec `pip` (comme `colorama`, `faker` ou `pillow`), une question se pose : **comment faire pour qu'une autre personne puisse exécuter ton projet sur sa machine sans devoir deviner quelles bibliothèques tu as utilisées ?**

On ne partage jamais les dossiers d'installation lourds. À la place, on crée un simple fichier texte nommé **`requirements.txt`**.

Ce fichier sert de **carte d'identité des dépendances** de ton projet : il liste le nom exact de chaque bibliothèque externe ainsi que sa version.


### **Les 2 Commandes Fondamentales à Maîtriser**

Toutes ces commandes s'exécutent directement dans ton **Terminal**.

#### 1. **Figer et sauvegarder tes dépendances** (`pip freeze`)

Une fois que tu as installé tes bibliothèques et que ton projet fonctionne, tu génères le fichier `requirements.txt` d'un seul coup grâce à la redirection `>` :

```bash
pip freeze > requirements.txt

```

SI tu ouvres ce fichier texte, tu y verras une liste claire :

```text
colorama==0.4.6
Faker==24.1.0
Pillow==10.2.0

```

> 💡 L'opérateur `==` garantit que toute personne qui télécharge ton projet installera **exactement la même version** que toi, évitant ainsi les bugs dus à des mises à jour incompatibles.

#### 2. **Installer toutes les dépendances d'un projet** (`pip install -r`)

Quand tu récupères le projet d'un collègue (ou un projet ouvert sur GitHub) qui contient un fichier `requirements.txt`, tu n'as pas besoin de taper `pip install` pour chaque bibliothèque une par une.

Il te suffit d'exécuter :

```bash
pip install -r requirements.txt

```

*(Le fan `-r` indique à `pip` de lire le fichier et de tout installer automatiquement).*

## 5.3 **<u>LES ENVIRONNEMENTS VIRTUELS</u> (`venv`)**

### **Pourquoi l'environnement virtuel est indispensable ?**

Jusqu'ici, quand tu exécutais `pip install colorama` ou `pip install pillow`, la bibliothèque s'installait **globalement** dans le Python principal de ton ordinateur.

**Le problème qui survient en pratique :**
Imagine que dans 6 mois tu travailles sur deux projets différents :

* **Projet A :** Nécessite la version 2.0 d'une bibliothèque.
* **Projet B :** Plus récent, nécessite la version 3.0 de la même bibliothèque.

Si tu restes en installation globale, installer la version 3.0 va écraser la version 2.0 et casser complètement le Projet A.

**La solution :**
Un **environnement virtuel** (`venv`) est un dossier isolé à l'intérieur de ton projet qui contient sa propre copie de Python et son propre dossier de bibliothèques. Tout ce que tu installes dedans reste **uniquement** dans ce projet.

```text
Mon_Projet/
├── .venv/               <-- L'environnement isolé (propre à ce projet)
├── main.py              <-- Ton code Python
└── requirements.txt     <-- Tes dépendances figées

```

---

### **Les 3 étapes du cycle de vie d'un `venv`**

Toutes ces commandes se tapent dans le **Terminal** à la racine de ton dossier de projet.

#### 1. **Créer l'environnement virtuel**

```bash
python -m venv .venv

```

*(Cette commande crée un dossier nommé `.venv` contenant ton Python isolé).*

#### 2. **Activer l'environnement virtuel**

Pour dire à ton terminal *"À partir de maintenant, utilise le Python du projet et pas celui de la machine"*, il faut l'activer :

* **Sous Windows (Invite de commandes CMD) :**
```cmd
.venv\Scripts\activate

```


* **Sous Windows (PowerShell) :**
```powershell
.venv\Scripts\Activate.ps1

```


* **Sous Linux / macOS / Git Bash :**
```bash
source .venv/bin/activate

```



> 🟢 **Signe de réussite :** Tu verras **`(.venv)`** s'afficher au tout début de la ligne de commande dans ton terminal !

#### 3. **Désactiver l'environnement virtuel**

Quand tu as fini de travailler sur ce projet et que tu veux fermer le canal isolé :

```bash
deactivate

```

---

### **Le Workflow Professionnel Complet**

Voici le cheminement exact que tu utiliseras à chaque fois que tu démarreras un nouveau projet Python :

1. **Créer le projet :** Tu crées ton dossier et tu l'ouvres dans le terminal.
2. **Isoler :** Tu crées le `venv` (`python -m venv .venv`) et tu l'actives.
3. **Travailler :** Tu installes tes bibliothèques (`pip install ...`).
4. **Figer :** Tu sauvegardes ta liste (`pip freeze > requirements.txt`).
5. **Terminer :** Tu désactives l'environnement (`deactivate`).
