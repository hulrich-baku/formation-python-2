### 4.1. <U>**L'OPÉRATEUR TERNAIRE**</U> (Expression Conditionnelle)
#### **Qu'est-ce que c'est ?**

L'opérateur ternaire permet d'assigner une valeur à une variable ou de retourner un résultat en fonction d'une condition, **en une seule ligne de code**.


#### **Syntaxe & Comparaison**

En Python, la syntaxe lit presque comme une phrase en anglais :

$$\text{valeur\_si\_vrai} \quad \mathbf{\text{if}} \quad \text{condition} \quad \mathbf{\text{else}} \quad \text{valeur\_si\_faux}$$

```python
# ❌ Méthode classique (4 lignes)
if note >= 10:
  statut = "Admis"
else:
  statut = "Ajourné"

# 🚀 Méthode Pythonique (1 seule ligne)
statut = "Admis" if note >= 10 else "Ajourné"

```

#### **Cas d'usage fréquents**

**1. Dans le retour d'une fonction (`return`) :**

```python
def obtenir_reduction(age: int) -> float:
  """Retourne le taux de réduction selon l'âge."""
  return 0.50 if age < 18 or age >= 65 else 0.0

```

**2. Valeurs par défaut personnalisées :**

```python
def saluer(nom: str | None = None) -> str:
  nom_affiche = nom if nom is not None else "Visiteur"
  return f"Bonjour, {nom_affiche} !"

```

> ⚠️ **Bonne pratique PEP 8 :** L'opérateur ternaire ne doit être utilisé que pour des conditions **simples**. N'imbriquez pas plusieurs ternaires (`a if c1 else b if c2 else c`), cela détruit la lisibilité.

## 4.2. <U>**LES COMPRÉHENSIONS DE LISTES ET DE DICTIONNAIRES**</U>
### **Qu'est-ce qu'une Compréhension ?**

Une **compréhension** permet de créer une nouvelle liste, un nouveau dictionnaire ou un nouveau jeu de données (*set*) à partir d'un itérable existant, **en une seule ligne de code concise et lisible**.

Elle remplace avantageusement les boucles `for` classiques associées à `.append()` ou au remplissage manuel.

---

### 1. **Les Compréhensions de Listes** (*List Comprehensions*)

#### A. **Transformation simple** (sans condition)

**Syntaxe :**

```python
[expression for element in itérable]

```

**Exemple :**

```python
fruits: list[str] = ["mangue", "orange", "avocat", "ananas", "banane"]

# On veut mettre la première lettre de chaque fruit en majuscule
# ❌ Méthode classique avec boucle for
fruits_majuscule: list[str] = []

for f in fruits:
  fruits_majuscule.append(f.capitalize())

# 🚀 Méthode Pythonique (Compréhension)
fruits_majuscule: list[str] = [f.capitalize() for f in fruits]
```


#### B. **Filtrage avec condition `if`** (À la fin)

Si l'on souhaite **filtrer** les éléments pour ne garder que ceux qui respectent un critère :

**Syntaxe :**

```python
[expression for element in itérable if condition]

```

**Exemple :**

```python
notes: list[float] = [8.5, 14.0, 9.0, 17.5, 12.0]

# Conserver uniquement les notes supérieures ou égales à 10.0
notes_validees: list[float] = [note for note in notes if note >= 10.0]
# Résultat : [14.0, 17.5, 12.0]

```


#### C. **Transformation avec `if / else`**  (Au début)

Si l'on souhaite appliquer une transformation **conditionnelle** (à l'aide d'un ternaire) sur tous les éléments :

**Syntaxe :**

```python
[valeur_si_vrai if condition else valeur_si_faux for element in itérable]

```

**Exemple :**

```python
notes: list[float] = [8.5, 14.0, 9.0, 17.5]

# Convertir en "Admis" ou "Rattrapage"
resultats: list[str] = [
    "Admis" if note >= 10.0 else "Rattrapage" for note in notes
]
# Résultat : ['Rattrapage', 'Admis', 'Rattrapage', 'Admis']

```


### 2. **Les Compréhensions de Dictionnaires** (*Dict Comprehensions*)

Elles permettent de construire un dictionnaire dynamiquement en précisant la paire `clé: valeur`.

**Syntaxe :**

```python
{cle_expression: valeur_expression for element in itérable}

```

**Exemple :**

```python
etudiants: list[str] = ["Hulrich", "Marc", "Alice"]

# Créer un dictionnaire associant chaque étudiant à la longueur de son nom
longueur_noms: dict[str, int] = {nom: len(nom) for nom in etudiants}
# Résultat : {'Hulrich': 7, 'Marc': 4, 'Alice': 5}

```

#### **Filtrage dans un dictionnaire** :

```python
catalogue_brut: dict[str, float] = {
    "Clavier": 45.0,
    "Ecran": 150.0,
    "Souris": 15.0,
}

# Filtrer pour ne garder que les produits chers (> 30.0)
produits_chers: dict[str, float] = {
    nom: prix for nom, prix in catalogue_brut.items() if prix > 30.0
}
# Résultat : {'Clavier': 45.0, 'Ecran': 150.0}

```

## 4.3. <u>**LES FONCTIONS ANONYMES (`LAMBDA`) & TRIS BASÉS SUR DES CLÉS**</u>

### 1. **Qu'est-ce qu'une Fonction `lambda` ?**

Une fonction **`lambda`** (ou fonction anonyme) est une fonction condensée définie **en une seule ligne**, sans nom officiel (`def`), destinée à être exécutée rapidement là où elle est définie.

#### **Syntaxe & Comparaison**

$$\mathbf{\text{lambda}} \quad \text{paramètres} \quad \mathbf{:} \quad \text{expression}$$

```python
# ❌ Avec def classique (3 lignes)
def carre_classique(x: float) -> float:
  return x**2

print(carre_classique(5)) # affiche 25


# 🚀 Avec lambda (1 ligne)
carre_lambda = lambda x: x**2

print(carre_lambda(5)) # affiche 25
```

> 💡 **Règle d'or :** Une fonction `lambda` ne contient pas le mot-clé `return`. L'expression située après le deux-points `:` est automatiquement évaluée et renvoyée.


### 2. **Cas d'Usage Principaux : Les Tris Complexes avec `.sort()` et `sorted()`**

Le cas d'usage le plus puissant des fonctions `lambda` en Python est de définir **une clé d'extraction (`key`)** pour trier des collections complexes (listes de tuples, listes de dictionnaires).

#### A. **Trier une liste de tuples**

Par défaut, Python trie les tuples selon le premier élément. Une fonction `lambda` permet de trier selon n'importe quel autre élément :

```python
# Liste d'étudiants: (Nom, Note)
etudiants: list[tuple[str, float]] = [
    ("Hulrich", 16.5),
    ("Alice", 18.0),
    ("Marc", 12.0),
]

# Trier par NOTE décroissante (du plus grand au plus petit)
etudiants_tries = sorted(etudiants, key=lambda student: student[1], reverse=True)
# Résultat : [('Alice', 18.0), ('Hulrich', 16.5), ('Marc', 12.0)]

```

#### B. **Trier une liste de dictionnaires**

Lors de la manipulation de données JSON ou d'API, on trie très fréquemment des listes de dictionnaires par une clé spécifique :

```python
from typing import TypedDict


class Produit(TypedDict):
  nom: str
  prix: float


catalogue: list[Produit] = [
    {"nom": "Écran", "prix": 250.0},
    {"nom": "Souris", "prix": 20.0},
    {"nom": "Clavier", "prix": 85.0},
]

# Trier le catalogue par PRIX croissant
catalogue.sort(key=lambda item: item["prix"])
# Le catalogue est maintenant trié : Souris (20.0), Clavier (85.0), Écran (250.0)

```
