
### 2. **La Structure de Base d'une Application**

Pour afficher une fenêtre à l'écran, il faut 3 éléments indissociables :

1. Configurer l'apparence générale (mode sombre ou clair).
2. Créer l'objet fenêtre (la racine de l'application).
3. Lancer la **boucle principale** (`mainloop()`) qui garde la fenêtre ouverte et attend les clics de l'utilisateur.

Voici le code minimal :

```python
import customtkinter as ctk

# 1. Configuration du thème visuel
ctk.set_appearance_mode("Dark")  # Options : "System", "Dark", "Light"
ctk.set_default_color_theme("blue")  # Options : "blue", "green", "dark-blue"

# 2. Création de la fenêtre principale
app = ctk.CTk()

# Configuration de la fenêtre
app.title("Ma Première Application")  # Titre dans la barre supérieure
app.geometry("400x300")  # Largeur x Hauteur en pixels

# 3. Lancement de la fenêtre (garde l'application ouverte)
app.mainloop()

```

### **Ce qu'il faut retenir** :

* **`app = ctk.CTk()`** : Crée la fenêtre principale.
* **`app.geometry("400x300")`** : Fixe la taille initiale (400 pixels de large, 300 pixels de haut).
* **`app.mainloop()`** : **Obligatoire à la fin du script.** Sans cette ligne, la fenêtre s'ouvrirait et se fermerait instantanément en une fraction de seconde !



## 🟢 <font color='#4cf'>**Point 2 : Afficher du texte (`CTkLabel`)**</font>

Un **`CTkLabel`** sert à afficher du texte (ou une image) dans ta fenêtre. C'est l'équivalent de ton `print()`, mais dans une interface graphique !

---

### 1. **Créer et placer un label simple**

Pour afficher un composant dans une fenêtre, on procède toujours en 2 temps :

1. **Créer le widget** en lui précisant où il se trouve (ici dans `app`) et son texte.
2. **Le placer à l'écran** grâce à la méthode `.pack()`.

```python
import customtkinter as ctk

app = ctk.CTk()
app.geometry("400x300")

# 1. Création du texte
mon_texte = ctk.CTkLabel(app, text="Bonjour et bienvenue !")

# 2. Placement du texte dans la fenêtre
mon_texte.pack(pady=20)  # pady=20 ajoute une marge verticale au-dessus et en dessous

app.mainloop()

```

### 2. **Personnaliser la Police, la Couleur et le Style**

Tu peux facilement changer l'apparence de ton texte avec le paramètre `font` (pour la police, la taille et le style) et `text_color` (pour la couleur).

#### A. **La police, la taille, le gras et l'italique (`font`)**

On passe un tuple à `font` : `(nom_police, taille, style)`.

```python
# Texte en gras (bold)
titre = ctk.CTkLabel(
    app,
    text="Titre Principal",
    font=("Arial", 24, "bold"),
)

# Texte en italique (italic)
sous_titre = ctk.CTkLabel(
    app,
    text="Sous-titre explicatif",
    font=("Consolas", 14, "italic"),
)

# Texte combiné (gras + italique)
remarque = ctk.CTkLabel(
    app,
    text="Attention : action importante",
    font=("Roboto", 12, "bold", "italic"),
)

```

#### B. **La couleur du texte (`text_color`)**

Tu peux utiliser des noms de couleurs en anglais (`"red"`, `"green"`, `"cyan"`) ou des codes hexadécimaux (`"#FF5733"`).

```python
label_colore = ctk.CTkLabel(
    app,
    text="Texte en vert brillant !",
    font=("Arial", 16, "bold"),
    text_color="#2ECC71",  # Vert
)

```


### **Exemple Complet À Tester**

Voici un script qui rassemble tous ces styles dans une seule fenêtre :

```python
import customtkinter as ctk

ctk.set_appearance_mode("Dark")

app = ctk.CTk()
app.title("Test des CTkLabel")
app.geometry("500x350")

# 1. Grand Titre en gras et en couleur
titre = ctk.CTkLabel(
    app,
    text="🎮 Bienvenue dans l'Application",
    font=("Helvetica", 22, "bold"),
    text_color="#3498DB",  # Bleu
)
titre.pack(pady=15)

# 2. Sous-titre explicatif
description = ctk.CTkLabel(
    app,
    text="Ceci est une démonstration du composant CTkLabel.",
    font=("Arial", 14),
)
description.pack(pady=10)

# 3. Message d'avertissement en italique
alerte = ctk.CTkLabel(
    app,
    text="* Assurez-vous d'avoir enregistré vos modifications.",
    font=("Arial", 12, "italic"),
    text_color="#E74C3C",  # Rouge
)
alerte.pack(pady=10)

app.mainloop()

```

## 🟢 <font color='#4cf'>**Point 3 : Le Champ de Saisie (`CTkEntry`)**</font>

Le **`CTkEntry`** permet à l'utilisateur de taper du texte sur une seule ligne (ex: nom d'utilisateur, mot de passe, âge, ville). C'est l'équivalent de la fonction `input()` du terminal !

---

### 1. **Créer un champ de saisie simple**

Comme pour le `CTkLabel`, on l'initialise et on le place à l'écran avec `.pack()` :

```python
import customtkinter as ctk

app = ctk.CTk()
app.geometry("400x300")

# Création du champ de saisie
champ_nom = ctk.CTkEntry(app)
champ_nom.pack(pady=20)

app.mainloop()

```

### 2. **Personnaliser le champ de saisie**

Tu peux styliser ton champ pour le rendre beaucoup plus agréable et clair :

#### A. **Le texte d'aide / d'exemple (`placeholder_text`)**

C'est le texte gris qui s'affiche automatiquement dans le champ avant que l'utilisateur ne commence à taper.

```python
champ = ctk.CTkEntry(app, placeholder_text="Saisis ton prénom ici...")

```

#### B. **Dimensions et coins arrondis (`width`, `height`, `corner_radius`)**

* **`width` / `height` :** Largeur et hauteur en pixels.
* **`corner_radius` :** Degré d'arrondi des angles (ex: `10` ou `20` pour un effet très arrondi).

```python
champ = ctk.CTkEntry(
    app,
    placeholder_text="Rechercher...",
    width=250,
    height=40,
    corner_radius=15,  # Coins très arrondis
)

```

#### C. **Les couleurs & bordures (`border_color`, `fg_color`, `text_color`)**

* **`fg_color` :** Couleur de fond du champ.
* **`border_color` / `border_width` :** Couleur et épaisseur de la bordure.
* **`text_color` :** Couleur du texte tapé par l'utilisateur.

```python
champ = ctk.CTkEntry(
    app,
    placeholder_text="Adresse e-mail",
    border_color="#3498DB",  # Bordure bleue
    border_width=2,
    fg_color="#1E1E1E",  # Fond sombre
    text_color="white",  # Texte tapé en blanc
)

```

#### D. **Masquer la saisie pour un mot de passe (`show`)**

En passant `show="*"`, les caractères tapés sont remplacés par des puces ou étoiles.

```python
champ_mdp = ctk.CTkEntry(
    app, placeholder_text="Mot de passe", show="*"
)

```

### 3. **Récupérer et modifier la valeur en Python (`.get()` et `.delete()`)**

Pour interagir avec ce que l'utilisateur a tapé :

* **`champ.get()` :** Renvoie la chaîne de caractères actuellement saisie dans le champ.
* **`champ.delete(0, "end")` :** Vide complètement le contenu du champ.


### **Exemple Complet À Tester**

Voici un script récapitulatif pour observer les différents styles :

```python
import customtkinter as ctk

ctk.set_appearance_mode("Dark")

app = ctk.CTk()
app.title("Test des CTkEntry")
app.geometry("400x350")

# 1. Label d'instruction
titre = ctk.CTkLabel(
    app, text="Formulaire d'exemple", font=("Arial", 18, "bold")
)
titre.pack(pady=15)

# 2. Champ classique avec placeholder
champ_pseudo = ctk.CTkEntry(
    app, placeholder_text="Nom d'utilisateur", width=250, corner_radius=10
)
champ_pseudo.pack(pady=10)

# 3. Champ stylisé avec bordure bleue
champ_email = ctk.CTkEntry(
    app,
    placeholder_text="Ex: exemple@email.com",
    width=250,
    border_color="#3498DB",
    border_width=2,
    corner_radius=10,
)
champ_email.pack(pady=10)

# 4. Champ mot de passe masqué
champ_mdp = ctk.CTkEntry(
    app, placeholder_text="Mot de passe", show="*", width=250, corner_radius=10
)
champ_mdp.pack(pady=10)

app.mainloop()

```

## 🟢 <font color='#4cf'>**Point 4 : Le Bouton (`CTkButton`)**</font>

Le **`CTkButton`** est l'élément interactif clé d'une interface graphique. Il permet de déclencher une fonction Python (enregistrer des données, calculer un résultat, valider un formulaire, etc.) lorsqu'on clique dessus.

---

### 1. **Créer un bouton simple lié à une fonction (`command`)**

Pour lier une fonction Python à un bouton, on utilise le paramètre **`command`**.

⚠️ **Attention :** On passe le nom de la fonction *sans* les parenthèses `()` (ex: `command=ma_fonction`, et non `command=ma_fonction()`).

```python
import customtkinter as ctk

app = ctk.CTk()
app.geometry("400x200")


# 1. Définition de la fonction qui sera exécutée au clic
def dire_bonjour():
  print("Le bouton a été cliqué !")


# 2. Création du bouton avec le paramètre command
bouton = ctk.CTkButton(app, text="Cliquez ici", command=dire_bonjour)
bouton.pack(pady=40)

app.mainloop()

```


### 2. **Connecter `CTkEntry`, `CTkButton` et `CTkLabel`**

C'est ici que la magie de l'interface graphique commence ! On récupère ce que l'utilisateur a écrit dans le champ `CTkEntry` grâce à `.get()`, puis on met à jour le texte du `CTkLabel` avec `.configure()`.

```python
import customtkinter as ctk

app = ctk.CTk()
app.geometry("400x300")


def valider():
  nom = champ_nom.get()  # Récupère le texte saisi
  if nom:
    label_resultat.configure(
        text=f"Bonjour, {nom} !", text_color="#2ECC71"
    )
  else:
    label_resultat.configure(text="Veuillez entrer un nom !", text_color="red")


champ_nom = ctk.CTkEntry(app, placeholder_text="Entrez votre nom")
champ_nom.pack(pady=10)

bouton = ctk.CTkButton(app, text="Valider", command=valider)
bouton.pack(pady=10)

label_resultat = ctk.CTkLabel(app, text="")
label_resultat.pack(pady=10)

app.mainloop()

```


### 3. **Personnaliser le style du bouton**

CustomTkinter offre un contrôle complet sur les couleurs, survol du curseur et formes du bouton :

#### A. **Les couleurs principales et de survol (`fg_color`, `hover_color`)**

* **`fg_color` :** Couleur par défaut du bouton.
* **`hover_color` :** Couleur qui apparaît lorsque le curseur passe au-dessus du bouton.

```python
bouton_vert = ctk.CTkButton(
    app,
    text="Sauvegarder",
    fg_color="#2ECC71",  # Vert
    hover_color="#27AE60",  # Vert plus foncé au survol
)

```

#### B. **Arrondi des angles (`corner_radius`)**

* `corner_radius=0` : Bouton rectangulaire strict.
* `corner_radius=10` : Coins légèrement arrondis.
* `corner_radius=30` : Bouton style "capsule" / "pill button".

```python
bouton_arrondi = ctk.CTkButton(
    app, text="Bouton Capsule", corner_radius=25, height=45
)

```

#### C. **Bouton avec fond transparent / contour (`border_width`, `border_color`)**

Pour créer un bouton style "Outline" (translucide avec une bordure colorée) :

```python
bouton_contour = ctk.CTkButton(
    app,
    text="Annuler",
    fg_color="transparent",  # Pas de fond
    border_color="#E74C3C",  # Bordure rouge
    border_width=2,
    text_color="#E74C3C",  # Texte rouge
    hover_color="#5B2C2D",  # Teinte sombre au survol
)

```


### **Exemple Complet À Tester**

Voici un script intégrant différents styles et une interaction complète :

```python
import customtkinter as ctk

ctk.set_appearance_mode("Dark")

app = ctk.CTk()
app.title("Démonstration des CTkButton")
app.geometry("450x400")

# --- FONCTION ---


def effacer_champs():
  champ_saisie.delete(0, "end")
  label_message.configure(text="Champ réinitialisé.", text_color="gray")


def afficher_message():
  texte = champ_saisie.get()
  if texte:
    label_message.configure(text=f"Enregistré : {texte}", text_color="#2ECC71")
  else:
    label_message.configure(text="Le champ est vide !", text_color="#E74C3C")


# --- COMPOSANTS ---

titre = ctk.CTkLabel(
    app, text="Gestionnaire de Saisie", font=("Arial", 20, "bold")
)
titre.pack(pady=15)

champ_saisie = ctk.CTkEntry(
    app, placeholder_text="Tapez quelque chose...", width=300, height=35
)
champ_saisie.pack(pady=10)

# Bouton d'action principal (Vert / Arrondi)
btn_valider = ctk.CTkButton(
    app,
    text="Valider la saisie",
    command=afficher_message,
    fg_color="#2ECC71",
    hover_color="#27AE60",
    corner_radius=15,
    width=200,
    height=40,
)
btn_valider.pack(pady=10)

# Bouton style Outline (Rouge / Transparent)
btn_effacer = ctk.CTkButton(
    app,
    text="Effacer",
    command=effacer_champs,
    fg_color="transparent",
    border_color="#E74C3C",
    border_width=2,
    text_color="#E74C3C",
    hover_color="#441919",
    corner_radius=15,
    width=200,
)
btn_effacer.pack(pady=10)

label_message = ctk.CTkLabel(
    app, text="", font=("Arial", 14, "italic"), text_color="gray"
)
label_message.pack(pady=15)

app.mainloop()

```


## 🟢 <font color='#4cf'>**Point 5 : Le Système de Disposition (*Layout Management*)**</font>

Jusqu'à présent, nous avons utilisé `.pack()` sans trop nous poser de questions. En réalité, Python propose **3 méthodes différentes** pour positionner les widgets dans une fenêtre : `.pack()`, `.grid()` et `.place()`.

---

### 1. **La méthode `.pack()` (Empilement simple)**

**Concept :** Il empile les éléments les uns au-dessus des autres (ou les uns à côté des autres). C'est idéal pour des interfaces simples et verticales.

#### **Propriétés clés** :

* **`pady=10` / `padx=10` :** Ajoute une marge externe verticale (haut/bas) ou horizontale (gauche/droite).
* **`side="top"` :** Direction de l'empilement (`"top"`, `"bottom"`, `"left"`, `"right"`).
* **`fill="x"` :** Étire le widget pour qu'il occupe toute la largeur disponible.

```python
label.pack(pady=10)  # Empilé verticalement avec une marge de 10px
bouton.pack(side="left", padx=5)  # Aligné à gauche

```


### 2. **La méthode `.grid()` (Grille en lignes et colonnes) — _La plus utilisée !_**

**Concept :** Il découpe la fenêtre en une grille invisible composée de **lignes** (`row`) et de **colonnes** (`column`). C'est le système le plus puissant pour structurer des formulaires ou des tableaux de bord.

```
       Column 0         Column 1
     ┌───────────────┬───────────────┐
Row 0│  CTkLabel     │  CTkEntry     │
     ├───────────────┼───────────────┤
Row 1│  CTkButton    │  CTkButton    │
     └───────────────┴───────────────┘

```

#### **Exemple de code avec `grid`** :

```python
import customtkinter as ctk

app = ctk.CTk()
app.geometry("400x200")

# Ligne 0 : Label à gauche (col 0), Champ à droite (col 1)
label_nom = ctk.CTkLabel(app, text="Nom :")
label_nom.grid(row=0, column=0, padx=10, pady=10)

entry_nom = ctk.CTkEntry(app, placeholder_text="Ton nom")
entry_nom.grid(row=0, column=1, padx=10, pady=10)

# Ligne 1 : Bouton centré ou aligné
btn_valider = ctk.CTkButton(app, text="Envoyer")
btn_valider.grid(row=1, column=0, columnspan=2, pady=20)  # Fusionne 2 colonnes

app.mainloop()

```


### 3. **La méthode `.place()` (Positionnement absolu) — *À titre informatif***

**Concept :** Il place le widget à des coordonnées exactes en pixels `(x, y)` par rapport au coin supérieur gauche de la fenêtre `(0, 0)`.

```
(0,0) ┌────────────────────────┐
      │  x=50, y=30            │
      │  ┌───────────┐         │
      │  │ CTkButton │         │
      │  └───────────┘         │
      └────────────────────────┘

```

#### **Exemple de code** :

```python
# Place le bouton exactement à 50px de la gauche et 80px du haut
bouton.place(x=50, y=80)

```

⚠️ **Pourquoi `.place()` est déconseillé dans la plupart des cas ?**

* Si l'utilisateur redimensionne la fenêtre ou change de résolution d'écran, les widgets **ne s'adaptent pas** et risquent d'être coupés ou mal alignés.
* Réservez `.place()` uniquement pour des cas très spécifiques (par exemple : placer un petit badge de notification au-dessus d'une icône).


## **Récapitulatif des 3 méthodes**

| Méthode | Rôle | Utilisation recommandée |
| --- | --- | --- |
| **`.pack()`** | Empile les éléments à la suite. | Interfaces simples, listes verticales rapides. |
| **`.grid()`** | Aligne en lignes et colonnes. | **Standard absolu** pour formulaires et applications complètes. |
| **`.place()`** | Positionne au pixel près `(x, y)`. | Usage rare (badges, superpositions spécifiques). |

<br/>

> 🚨 **Règle d'or :** Dans une même fenêtre ou un même conteneur, **ne mélange jamais `.pack()` et `.grid()`**, sous peine de faire planter ton application !

Dans cette étape, nous allons apprendre à dépasser la simple fenêtre "plate" pour construire de véritables architectures d'applications professionnelles avec un **découpage en blocs (Frames)**, des **grilles adaptatives**, une **barre latérale (Sidebar)** et une **zone principale**.

## 🟡 <font color='#4cf'>**Point 1 : Les Conteneurs (`CTkFrame`)**</font>

Dans une application moderne, on ne place presque jamais tous les widgets directement sur la fenêtre principale `app`. On découpe la fenêtre en plusieurs "sous-boîtes" appelées **`CTkFrame`**.

---

### 1. **Qu'est-ce qu'une `CTkFrame` ?**

Une `CTkFrame` est un conteneur rectangulaire invisible (ou avec un fond de couleur) dans lequel tu peux regrouper d'autres widgets (`CTkLabel`, `CTkButton`, etc.).

```
┌─────────────────────────────────────────────────────────┐
│ Fenêtre Principale (app)                                │
│                                                         │
│  ┌────────────────────────┐  ┌───────────────────────┐  │
│  │ CTkFrame (Gauche)      │  │ CTkFrame (Droite)     │  │
│  │                        │  │                       │  │
│  │  • CTkButton 1         │  │  • CTkLabel           │  │
│  │  • CTkButton 2         │  │  • CTkEntry           │  │
│  └────────────────────────┘  └───────────────────────┘  │
└─────────────────────────────────────────────────────────┘

```


### 2. **Créer et Styliser une `CTkFrame`**

Pour créer une frame, on la rattache à `app`. Ensuite, lorsqu'on crée un widget à l'intérieur de cette frame, **on passe la frame comme premier argument** au lieu de `app` !

```python
import customtkinter as ctk

app = ctk.CTk()
app.geometry("500x300")

# 1. Création d'une Frame avec un fond spécifique et des coins arrondis
ma_carte = ctk.CTkFrame(
    app, fg_color="#2B2B2B", corner_radius=15, border_width=2, border_color="#3498DB"
)
ma_carte.pack(padx=20, pady=20, fill="both", expand=True)

# 2. Ajout de widgets À L'INTÉRIEUR de la frame (note le 'ma_carte' au lieu de 'app')
titre_carte = ctk.CTkLabel(
    ma_carte, text="Carte d'Information", font=("Arial", 16, "bold")
)
titre_carte.pack(pady=10)

btn_carte = ctk.CTkButton(ma_carte, text="Action")
btn_carte.pack(pady=10)

app.mainloop()

```

### **Propriétés importantes de `CTkFrame`** :

* **`fg_color` :** Couleur de fond de la carte.
* **`corner_radius` :** Arrondi des bordures.
* **`border_width` & `border_color` :** Pour créer un contour élégant autour du bloc.


## 🟡 <font color='#4cf'>**Point 2 : Grilles Réactives (`grid`)**</font>

Par défaut, quand tu utilises `.grid()`, les colonnes et les lignes se resserrent au maximum autour de la taille de tes widgets. Si l'utilisateur agrandit ou redimensionne la fenêtre, les widgets restent figés au centre ou dans un coin sans s'étendre.

Pour rendre ton interface **réactive (*responsive*)**, il faut indiquer à la fenêtre comment distribuer l'espace supplémentaire lors du redimensionnement.

---

### 1. **La méthode `grid_columnconfigure()` et `grid_rowconfigure()`**

Ces deux méthodes s'appliquent sur le conteneur parent (`app` ou une `CTkFrame`). Elles permettent de définir le **poids (*weight*)** de chaque ligne ou colonne.

* **`weight=0` (Par défaut) :** La ligne/colonne garde une taille fixe minimale (celle de son contenu).
* **`weight=1` (ou plus) :** La ligne/colonne s'étire dynamiquement pour occuper tout l'espace disponible.

```python
# La colonne 0 s'étire dynamiquement avec la fenêtre
app.grid_columnconfigure(0, weight=1)

# La ligne 1 s'étire dynamiquement
app.grid_rowconfigure(1, weight=1)

```


### 2. **Le paramètre `sticky` dans `.grid()`**

Définir un `weight` ne suffit pas : il faut aussi indiquer au widget **dans quelle direction s'étirer** au sein de sa cellule de grille. On utilise le paramètre `sticky` avec les points cardinaux en anglais :

* **`n`** (North / Haut)
* **`s`** (South / Bas)
* **`e`** (East / Droite)
* **`w`** (West / Gauche)

#### **Combinaisons courantes** :

* **`sticky="ew"` :** Étire le widget sur toute la **largeur** (gauche-droite).
* **`sticky="ns"` :** Étire le widget sur toute la **hauteur** (haut-bas).
* **`sticky="nsew"` :** Étire le widget dans **toutes les directions** pour remplir à 100 % sa case.


### **Exemple : Une grille responsive à 2 colonnes**

Voici un exemple où la colonne de droite s'étire quand on redimensionne la fenêtre, tandis que la colonne de gauche (menu) conserve sa largeur :

```python
import customtkinter as ctk

app = ctk.CTk()
app.geometry("500x300")

# Configuration de la grille de la fenêtre :
# La colonne 0 reste fixe (weight=0 par défaut)
# La colonne 1 prend tout l'espace restant (weight=1)
app.grid_columnconfigure(1, weight=1)
app.grid_rowconfigure(0, weight=1)

# Colonne 0 : Menu latéral (Fixe)
menu = ctk.CTkFrame(app, width=140, corner_radius=0)
menu.grid(row=0, column=0, sticky="nsew")

btn1 = ctk.CTkButton(menu, text="Accueil")
btn1.pack(pady=10, padx=10)

# Colonne 1 : Zone de contenu principale (S'étire !)
contenu = ctk.CTkFrame(app, fg_color="#2B2B2B")
contenu.grid(row=0, column=1, padx=10, pady=10, sticky="nsew")

titre = ctk.CTkLabel(
    contenu, text="Zone Principale", font=("Arial", 18, "bold")
)
titre.pack(pady=20)

app.mainloop()

```

## 🟡 <font color='#4cf'>**Point 3 : Créer une Sidebar et une Navbar**</font>

Dans les applications modernes (comme Discord, VS Code ou Spotify), la mise en page repose sur une structure standard très claire :

* **Sidebar (Barre latérale) :** Située à gauche, elle contient le menu de navigation, les options principales ou le profil.
* **Navbar (Barre supérieure) :** Située en haut, elle affiche le titre de la page actuelle, une barre de recherche ou le changement de thème.
* **Zone Centrale (Main Content) :** L'espace principal qui affiche le contenu dynamique.

---

### **La Structure de la Grille (2x2)**

Pour assembler ces éléments sans chevauchement, on divise la fenêtre principale `app` en une grille de **2 lignes** et **2 colonnes** :

```
             Col 0 (Fixe)            Col 1 (S'étire weight=1)
          ┌──────────────────────┬─────────────────────────────┐
Row 0     │                      │  Navbar (CTkFrame)          │
(Fixe)    │  Sidebar             ├─────────────────────────────┤
          │  (CTkFrame)          │                             │
Row 1     │                      │  Zone Centrale              │
(weight=1)│                      │  (CTkFrame)                 │
          └──────────────────────┴─────────────────────────────┘

```


### **Exemple de Code : Structure Complete**

Voici comment construire ce squelette propre et responsive :

```python
import customtkinter as ctk

ctk.set_appearance_mode("Dark")

app = ctk.CTk()
app.title("Application Modern Layout")
app.geometry("700x450")

# --- 1. CONFIGURATION DE LA GRILLE PRINCIPALE ---
# La colonne 1 (zone de contenu) s'étire en largeur
app.grid_columnconfigure(1, weight=1)

# La ligne 1 (zone centrale) s'étire en hauteur
app.grid_rowconfigure(0, weight=0)  # Navbar : hauteur fixe
app.grid_rowconfigure(1, weight=1)  # Contenu principal : s'étire

# --- 2. SIDEBAR (Barre Latérale - Colonne 0, s'étend sur les lignes 0 et 1) ---
sidebar = ctk.CTkFrame(app, width=160, corner_radius=0)
sidebar.grid(row=0, column=0, rowspan=2, sticky="nsew")

# Éléments de la Sidebar
logo = ctk.CTkLabel(
    sidebar, text="⚡ myApp", font=("Helvetica", 20, "bold"), text_color="#3498DB"
)
logo.pack(pady=20, padx=10)

btn_accueil = ctk.CTkButton(sidebar, text="🏠 Accueil", corner_radius=8)
btn_accueil.pack(pady=10, padx=15, fill="x")

btn_profil = ctk.CTkButton(
    sidebar,
    text="👤 Profil",
    fg_color="transparent",
    text_color="white",
    hover_color="#333333",
)
btn_profil.pack(pady=5, padx=15, fill="x")

# --- 3. NAVBAR (Barre Supérieure - Ligne 0, Colonne 1) ---
navbar = ctk.CTkFrame(app, height=50, corner_radius=0, fg_color="#1E1E1E")
navbar.grid(row=0, column=1, sticky="ew", padx=10, pady=(10, 0))

titre_page = ctk.CTkLabel(
    navbar, text="Tableau de Bord", font=("Arial", 16, "bold")
)
titre_page.pack(side="left", padx=15, pady=10)

# --- 4. ZONE DE CONTENU PRINCIPAL (Ligne 1, Colonne 1) ---
main_content = ctk.CTkFrame(app, fg_color="#2B2B2B", corner_radius=10)
main_content.grid(row=1, column=1, sticky="nsew", padx=10, pady=10)

label_bienvenue = ctk.CTkLabel(
    main_content,
    text="Bienvenue dans votre espace de travail !",
    font=("Arial", 14),
)
label_bienvenue.pack(pady=40)

app.mainloop()

```


### **Les points clés de ce code** :

1. **`rowspan=2` sur la Sidebar :** Permet à la barre latérale de s'étendre sur toute la hauteur de la fenêtre en couvrant à la fois la Ligne 0 et la Ligne 1.
2. **`side="left"` dans la Navbar :** Aligne les éléments horizontalement (ex: le titre de la page est collé sur la gauche de la barre supérieure).
3. **`fill="x"` sur les boutons :** Permet aux boutons de la Sidebar d'occuper toute la largeur disponible de la barre latérale.

Dans cette étape, nous allons rendre tes interfaces graphiques vivantes. Tu vas apprendre à synchroniser l'affichage automatiquement grâce aux variables dynamiques, à réagir aux interactions du clavier ou de la souris, et à valider les saisies des utilisateurs.

## 🟠 <font color='#4cf'>**Point 1 : Les Variables Dynamiques (`StringVar`, `IntVar`, `BooleanVar`)**</font>

Dans un script classique, pour modifier la valeur d'un widget, on utilise `.configure(text=...)`. C'est très bien pour des actions ponctuelles, mais pour synchroniser plusieurs widgets entre eux, Python et CustomTkinter proposent des **Variables Tkinter Spéciales**.

Ces variables sont "observées" par l'interface : dès que leur valeur change en Python, **tous les widgets reliés à cette variable se mettent à jour instantanément à l'écran**.

---

### 1. **Les 3 Types de Variables Principales**

* **`ctk.StringVar()` :** Pour stocker une chaîne de caractères (texte).
* **`ctk.IntVar()` :** Pour stocker un nombre entier (`0`, `1`, `42`...).
* **`ctk.BooleanVar()` :** Pour stocker un booléen (`True` ou `False`).


### 2. **Méthodes clés : `.get()` et `.set()`**

On ne lit et ne modifie pas ces variables avec l'opérateur `=` classique. On utilise deux méthodes obligatoires :

* **`variable.get()` :** Récupère la valeur actuelle de la variable.
* **`variable.set(nouvelle_valeur)` :** Modifie la valeur et met à jour les widgets rattachés.

```python
import customtkinter as ctk

app = ctk.CTk()

# 1. Création de la variable dynamique
texte_dynamique = ctk.StringVar(value="Valeur initiale")

# 2. Lecture et Modification
print(texte_dynamique.get())  # Affiche: "Valeur initiale"
texte_dynamique.set("Nouvelle valeur !")  # Met à jour automatiquement la variable

```

### 3. **Réaction automatique avec `.trace_add()`**

Tu peux demander à Python d'exécuter une fonction **à chaque fois que la valeur de la variable change**, sans même avoir besoin de cliquer sur un bouton !

```python
import customtkinter as ctk

app = ctk.CTk()
app.geometry("400x250")

var_pseudo = ctk.StringVar()


# Fonction déclenchée à CHAQUE lettre tapée
def au_changement(*args):
  valeur = var_pseudo.get()
  label_compteur.configure(text=f"Nombre de caractères : {len(valeur)}")


# On attache la fonction à la variable ('write' = quand la variable est modifiée)
var_pseudo.trace_add("write", au_changement)

# Le champ de saisie est directement lié à var_pseudo
entry = ctk.CTkEntry(
    app, textvariable=var_pseudo, placeholder_text="Tape ton pseudo..."
)
entry.pack(pady=20)

label_compteur = ctk.CTkLabel(app, text="Nombre de caractères : 0")
label_compteur.pack(pady=10)

app.mainloop()

```



## 🟠 <font color='#4cf'>**Point 2 : Événements Clavier et Souris (`.bind`)**</font>

Jusqu'à présent, nous avons utilisé le paramètre `command=` pour déclencher une action au clic sur un `CTkButton`. Cependant, pour créer une interface fluide, il est souvent indispensable de capter d'autres interactions :

* Valider un formulaire en appuyant sur la touche **Entrée** (`<Return>`).
* Fermer une fenêtre avec la touche **Échap** (`<Escape>`).
* Réagir au double-clic ou au passage de la souris sur un composant.

C'est là qu'intervient la méthode **`.bind()`**.

---

### 1. **La Syntaxe Générale de `.bind()`**

La méthode `.bind()` prend deux arguments principaux :

1. **L'événement cible** (écrit sous forme de chaîne entre crochets, ex: `"<Return>"`).
2. **La fonction callback** qui sera exécutée lorsque l'événement se produit.

```python
widget.bind("<NomDeLEvenement>", nom_de_la_fonction)

```

⚠️ **Règle fondamentale des callbacks avec `.bind()` :**
Contrairement aux fonctions utilisées avec `command=`, la fonction appelée par `.bind()` **reçoit automatiquement un paramètre `event**` transmis par Tkinter (qui contient des détails sur la touche pressée ou les coordonnées de la souris).

```python
# La fonction DOIT accepter l'argument 'event'
def sur_touche_presser(event):
  print(f"Touche pressée : {event.keysym}")

```


### 2. **Les Événements Clavier les Plus Courants**

| Code de l'Événement | Description |
| --- | --- |
| **`"<Return>"`** | Appui sur la touche **Entrée**. |
| **`"<Escape>"`** | Appui sur la touche **Échap**. |
| **`"<Key>"`** | N'importe quelle touche du clavier pressée. |
| **`"<Control-s>"`** | Combinaison de touches (ex: Raccourci Ctrl + S). |


### 3. **Les Événements Souris les Plus Courants**

| Code de l'Événement | Description |
| --- | --- |
| **`"<Button-1>"`** | Clic gauche de la souris. |
| **`"<Double-Button-1>"`** | Double-clic gauche. |
| **`"<Enter>"`** | La souris entre au-dessus du widget (survol). |
| **`"<Leave>"`** | La souris quitte la surface du widget. |


### **Exemple : Valider un champ avec la touche "Entrée"**

Voici un cas d'usage très classique : l'utilisateur tape du texte dans un `CTkEntry` et appuie directement sur la touche **Entrée** pour valider sans avoir à cliquer sur le bouton.

```python
import customtkinter as ctk

app = ctk.CTk()
app.geometry("400x250")


# 1. Fonction callback qui reçoit l'argument obligatoire 'event'
def valider_saisie(event):
  texte = entry_recherche.get()
  if texte:
    label_status.configure(
        text=f"Recherche lancée : '{texte}'", text_color="#2ECC71"
    )
  else:
    label_status.configure(
        text="Veuillez saisir un terme !", text_color="#E74C3C"
    )


label_titre = ctk.CTkLabel(
    app, text="Appuie sur 'Entrée' pour chercher", font=("Arial", 16, "bold")
)
label_titre.pack(pady=15)

entry_recherche = ctk.CTkEntry(
    app, placeholder_text="Tape ton texte...", width=250
)
entry_recherche.pack(pady=10)

# 2. Liaison de la touche Entrée au champ de saisie
entry_recherche.bind("<Return>", valider_saisie)

label_status = ctk.CTkLabel(app, text="")
label_status.pack(pady=15)

app.mainloop()

```


### **Astuce : Lier un événement à toute la fenêtre**

Si tu souhaites qu'un raccourci clavier fonctionne peu importe où se trouve le curseur dans l'application, applique le `.bind()` directement sur la fenêtre principale `app` :

```python
# Fermer l'application en appuyant sur Échap
app.bind("<Escape>", lambda event: app.destroy())

```

## 🟠 <font color='#4cf'>**Point 3 : Validation et États des Widgets (`disabled` / `normal`)**</font>

Pour offrir une expérience utilisateur fluide et éviter les erreurs de saisie, il est essentiel de :

1. **Valider les données** saisies par l'utilisateur (vérifier la présence d'un `@` dans un e-mail, s'assurer qu'un mot de passe n'est pas trop court, etc.).
2. **Contrôler l'état des widgets** avec la propriété `state` (`"normal"` ou `"disabled"`). Par exemple : désactiver le bouton d'envoi tant que tous les champs obligatoires ne sont pas correctement remplis.

---

### 1. **La propriété `state` (`"normal"` vs `"disabled"`)**

Tous les widgets interactifs (`CTkButton`, `CTkEntry`, `CTkCheckBox`, etc.) possèdent un paramètre `state` :

* **`state="normal"` :** Le widget est actif, cliquable et réactif.
* **`state="disabled"` :** Le widget est grisé, non cliquable et ne reçoit plus les événements.

#### **Modifier l'état dynamiquement avec `.configure()`** :

```python
# Activer un bouton
mon_bouton.configure(state="normal")

# Désactiver un bouton (le rend grisé)
mon_bouton.configure(state="disabled")

```


### 2. **Validation en temps réel avec `StringVar` et `trace_add`**

En combinant une `StringVar` sur un `CTkEntry` et la méthode `.trace_add()`, on peut vérifier la validité de la saisie à chaque caractère tapé et déverrouiller le bouton de validation automatiquement.

#### **Exemple : Exiger un e-mail valide pour débloquer le bouton**

```python
import customtkinter as ctk

app = ctk.CTk()
app.geometry("400x250")

var_email = ctk.StringVar()


def verifier_email(*args):
  email = var_email.get()

  # Condition simple : contient un '@' et un '.' avec au moins 5 caractères
  if "@" in email and "." in email and len(email) >= 5:
    btn_valider.configure(state="normal", fg_color="#2ECC71")  # Vert
    label_erreur.configure(text="")
  else:
    btn_valider.configure(state="disabled", fg_color="gray")  # Grisé
    label_erreur.configure(
        text="Adresse e-mail invalide", text_color="#E74C3C"
    )


# Champ lié à la variable
entry_email = ctk.CTkEntry(
    app,
    textvariable=var_email,
    placeholder_text="Entrez votre e-mail...",
    width=250,
)
entry_email.pack(pady=15)

# Déclenche la vérification à chaque modification du texte
var_email.trace_add("write", verifier_email)

label_erreur = ctk.CTkLabel(app, text="")
label_erreur.pack(pady=5)

# Bouton initialement DÉSACTIVÉ (state="disabled")
btn_valider = ctk.CTkButton(
    app, text="Continuer", state="disabled", fg_color="gray"
)
btn_valider.pack(pady=15)

app.mainloop()

```


### **Application Pratique : Validation de Mot de Passe**

Voici un cas très courant où l'on valide la longueur minimale d'un mot de passe en temps réel :

```python
import customtkinter as ctk

app = ctk.CTk()
app.geometry("400x280")

var_mdp = ctk.StringVar()


def verifier_mdp(*args):
  mdp = var_mdp.get()
  longueur = len(mdp)

  if longueur == 0:
    label_indic.configure(
        text="Le mot de passe doit faire 8 caractères min.", text_color="gray"
    )
    btn_connexion.configure(state="disabled")
  elif longueur < 8:
    label_indic.configure(
        text=f"Trop court ({longueur}/8)", text_color="#E74C3C"
    )
    btn_connexion.configure(state="disabled")
  else:
    label_indic.configure(text="Longueur valide !", text_color="#2ECC71")
    btn_connexion.configure(state="normal")


entry_mdp = ctk.CTkEntry(
    app,
    textvariable=var_mdp,
    show="*",
    placeholder_text="Mot de passe",
    width=250,
)
entry_mdp.pack(pady=15)

var_mdp.trace_add("write", verifier_mdp)

label_indic = ctk.CTkLabel(
    app,
    text="Le mot de passe doit faire 8 caractères min.",
    font=("Arial", 12),
    text_color="gray",
)
label_indic.pack(pady=5)

btn_connexion = ctk.CTkButton(
    app, text="Se connecter", state="disabled", width=250
)
btn_connexion.pack(pady=15)

app.mainloop()

```
