## 1. **RAPPEL : Arguments Positionnels vs Nommés**

En Python, il existe deux manières fondamentales de passer des valeurs (arguments) à une fonction lors de son appel : le mode **positionnel** et le mode **nommé** (ou par mot-clé).

* **Le mode positionnel :** Python se base uniquement sur l'ordre dans lequel les arguments sont écrits. La première valeur va dans le premier paramètre, la deuxième dans le deuxième, et ainsi de suite.
* **Le mode nommé :** On spécifie explicitement le nom du paramètre suivi de sa valeur (`nom_parametre=valeur`). L'ordre n'a alors plus aucune importance, ce qui sécurise la lecture du code.

```python
def creer_colis(nom, poids, fragile):
    print(f"Colis : {nom} | Poids : {poids}kg | Fragile : {fragile}")

# 1. APPEL POSITIONNEL
# "Miroir" va dans nom, 4.5 dans poids, et True dans fragile.
# Si on inverse l'ordre, les données seront affectées aux mauvais paramètres.
creer_colis("Miroir", 4.5, True)

# 2. APPEL NOMMÉ
# On lie chaque valeur à son paramètre. L'ordre peut être complètement modifié.
creer_colis(fragile=True, nom="Miroir", poids=4.5)

```

## 📦 2. **Dépaquetage avec `*args` et `**kwargs`**

Il arrive souvent qu'on ne connaisse pas à l'avance le nombre exact d'arguments qu'un utilisateur va envoyer à une fonction (par exemple, ajouter un nombre indéterminé d'articles dans un panier). C'est là qu'interviennent `*args` et `**kwargs`.

#### A. **Les arguments positionnels arbitraires : `*args`**

Le symbole `*` placé devant un nom de paramètre (souvent nommé `args` par convention) indique à Python de capturer toutes les valeurs positionnelles supplémentaires envoyées et de les regrouper automatiquement dans un **tuple** (une collection ordonnée non modifiable).

```python
def lister_produits(responsable, *args):
    print(f"Responsable : {responsable}")
    print(f"Structure de args : {type(args)}")  # Affiche <class 'tuple'>
    
    # On parcourt le tuple comme une liste classique
    for produit in args:
        print(f"- {produit}")

# "Baku" va dans responsable. Tout le reste ("Table", "Chaise", "Armoire") est capturé par *args.
lister_produits("Baku", "Table", "Chaise", "Armoire")

```

### B. **Les arguments nommés arbitraires : `**kwargs`**

Le double astérisque `**` (associé au terme `kwargs` pour *Keyword Arguments*) fait la même chose, mais pour les arguments nommés. Il capture toutes les variables passées sous la forme `cle=valeur` et les stocke dans un **dictionnaire**.

```python
def inspecter_caracteristiques(**kwargs):
    print(f"Structure de kwargs : {type(kwargs)}")  # Affiche <class 'dict'>
    
    # On parcourt le dictionnaire avec la méthode .items()
    for cle, valeur in kwargs.items():
        print(f"{cle} : {valeur}")

# Chaque argument nommé devient une paire clé/valeur dans le dictionnaire kwargs
inspecter_caracteristiques(couleur="Rouge", hauteur="2m", materiau="Bois")

```

## 🔏 3. **Les Paramètres Restrictifs : `/` et `*`**

Par défaut, Python permet de mélanger les appels positionnels et nommés pour la plupart des paramètres. Cependant, en ingénierie logicielle, il est parfois indispensable d'imposer des contraintes strictes pour éviter les erreurs d'utilisation ou figer une API.

* **La barre oblique `/` (Positional-Only) :** Elle agit comme une frontière. Tout paramètre situé à **gauche** du `/` doit obligatoirement être passé de manière positionnelle. Il est strictement interdit d'utiliser son nom lors de l'appel.
* **L'astérisque `*` (Keyword-Only) :** Tout paramètre situé à **droite** du `*` doit obligatoirement être passé de manière nommée (`cle=valeur`). Il est strictement interdit de le passer de manière positionnelle.

```python
def configurer_stock(emplacement, /, section, *, securise):
    """
    - 'emplacement' est avant le / : uniquement positionnel.
    - 'section' est entre le / et le * : flexible (positionnel ou nommé).
    - 'securise' est après le * : uniquement nommé.
    """
    print(f"Lieu: {emplacement} | Sec: {section} | Sécurisé: {securise}")

# APPEL VALIDE : respect parfait des frontières
configurer_stock("Zone-A", "Étagère-1", securise=True)

```

#### Exemples d'erreurs d'appels provoquées par les restrictions :

```python
# Lève une TypeError: 'emplacement' must be passed as a positional argument
configurer_stock(emplacement="Zone-A", section="Étagère-1", securise=True)

# Lève une TypeError: takes 2 positional arguments but 3 were given (True a été pris pour un positionnel)
configurer_stock("Zone-A", "Étagère-1", True)

```

### 🏛️ **L'Ordre Universel des Paramètres**

Pour mélanger tous ces types de paramètres, la signature de la fonction doit obligatoirement respecter l'ordre suivant (de gauche à droite) :

$$\text{Positionnels purs } (/) \rightarrow \text{ Flexibles } \rightarrow \text{ Arbitraires positionnels } (*args) \rightarrow \text{ Nommés purs } (*) \rightarrow \text{ Arbitraires nommés } (**kwargs)$$

> **⚠️ Règle d'or :**
> 1. Les paramètres positionnels (purs ou flexibles) se placent toujours **avant** `*args`.
> 2. Les paramètres nommés (purs ou avec valeurs par défaut) se placent toujours **après** `*args` et **avant** `**kwargs`.
> 3. `**kwargs` se trouve **toujours** en toute fin de signature.
> 
>

### 📝 Cas Pratique : La Signature Ultime

Voici une fonction qui combine l'intégralité des types de paramètres possibles en respectant l'ordre syntaxique :

```python
def super_generateur(pos_uniquement, /, flexible, *args, nom_uniquement, **kwargs):
    print(f"1. Positionnel pur (gauche de /) : {pos_uniquement}")
    print(f"2. Flexible (positionnel ou nommé) : {flexible}")
    print(f"3. args (tuple des positionnels restants) : {args}")
    print(f"4. Nommé pur (droite de *args) : {nom_uniquement}")
    print(f"5. kwargs (dict des nommés restants) : {kwargs}")

```

#### 🟢 Exemple d'appel valide :

```python
super_generateur(
    "Valeur 1",                    # Va dans 'pos_uniquement'
    "Valeur 2",                    # Va dans 'flexible'
    "Extra A", "Extra B",          # Capturés par '*args'
    nom_uniquement="Valeur Oblig", # Va dans 'nom_uniquement'
    option1="Oui", option2="Non"   # Capturés par '**kwargs'
)

```

#### 🔴 Exemples d'appels invalides :

```python
# ERREUR : Tentative de nommer le premier argument (qui est positionnel pur)
super_generateur(pos_uniquement="Valeur 1", flexible="Valeur 2", nom_uniquement="Ok")

# ERREUR : 'nom_uniquement' est passé de façon positionnelle (Python le cherche après *args)
super_generateur("V1", "V2", "Extra A", "Valeur Oblig")

```

## **FONDATIONS FONCTIONNELLES : FONCTIONS DE PREMIÈRE CLASSE & CLOSURES**

Avant d'attaquer la syntaxe des décorateurs, il est indispensable de comprendre deux mécanismes avancés de Python : le fait qu'une fonction soit un objet manipulable, et la capacité d'une fonction à capturer son environnement (la fermeture ou *closure*).

---

### 1. **Les Fonctions comme Objets de Première Classe**

En Python, les fonctions sont des objets de "première classe" (*First-Class Citizens*). Cela signifie qu'elles ont le même statut que n'importe quelle autre variable (comme un entier, une chaîne ou une liste).

On peut donc :

* Stocker une fonction dans une variable.
* Passer une fonction en argument d'une autre fonction.
* Retourner une fonction depuis une autre fonction.

#### **Stocker et renommer une fonction**

```python
def saluer():
    print("Bonjour !")

# On affecte l'objet fonction à une nouvelle variable (sans mettre les parenthèses)
dire_bonjour = saluer 

# On peut maintenant l'appeler via sa nouvelle variable
dire_bonjour()  # Affiche: Bonjour !

```

#### **Passer une fonction en paramètre**

```python
def executer_operation(fonction_a_executer, valeur):
    # La fonction passée en argument est appelée ici dynamiquement
    return fonction_a_executer(valeur)

def au_carre(x):
    return x ** 2

# On passe l'objet fonction 'au_carre' en argument
resultat = executer_operation(au_carre, 5)
print(resultat)  # Affiche: 25

```

---

### 2. **Les Fonctions Imbriquées**

Python permet de définir une fonction à l'intérieur d'une autre fonction. La fonction interne n'est alors visible et accessible que depuis l'espace local de la fonction externe.

```python
def fonction_externe(message):
    
    # Définition de la fonction interne
    def fonction_interne():
        print(f"Message reçu de l'externe : {message}")
    
    # Appel de la fonction interne depuis l'externe
    fonction_interne()

fonction_externe("Hello World")

```

---

### 3. **Le Mécanisme des Closures** (Fermetures)

Une **closure** se produit lorsqu'une fonction interne fait référence à des variables définies dans le scope de la fonction externe, et que la fonction externe **renvoie** cette fonction interne (au lieu de l'exécuter).

La fonction renvoyée "se souvient" et garde l'accès aux variables locales de la fonction parente, même après que la fonction parente a terminé son exécution.

```python
def creer_multiplicateur(facteur):
    # 'facteur' est une variable locale à creer_multiplicateur
    
    def multiplier(nombre):
        # La fonction interne utilise la variable 'facteur' de l'externe
        return nombre * facteur
    
    # On RETOURNE l'objet fonction multiplier sans l'exécuter
    return multiplier

# 'doubler' devient une fonction qui a mémorisé le facteur 2
doubler = creer_multiplicateur(2)

# 'tripler' devient une fonction qui a mémorisé le facteur 3
tripler = creer_multiplicateur(3)

print(doubler(10))  # Affiche: 20
print(tripler(10))  # Affiche: 30

```

Par définition, une **closure** est une fonction interne qui mémorise des variables de la fonction externe même après que cette dernière ait fini son exécution.

## ⚙️ **LE SUCRE SYNTAXIQUE & LA MÉCANIQUE DU `@`**

Maintenant que vous maîtrisez les closures (les fonctions qui créent et renvoient d'autres fonctions), vous avez toutes les clés pour comprendre ce qu'est réellement un **décorateur**.

---

### 1. **Qu'est-ce qu'un Décorateur ?**

Un **décorateur** est tout simplement une fonction qui prend une autre fonction en paramètre, étend ou modifie son comportement sans toucher à son code source, et renvoie cette nouvelle fonction modifiée.

Avant d'utiliser le symbole `@`, voyons comment Python réalise cette opération manuellement.

```python
# 1. Le décorateur : il prend une fonction 'f' en paramètre
def mon_decorateur(f):
    # Il crée une fonction interne (enveloppe / wrapper)
    def enveloppe():
        print("--- Avant l'exécution de la fonction ---")
        f()  # Appel de la fonction d'origine
        print("--- Après l'exécution de la fonction ---")
    
    # Il retourne la fonction interne modifiée
    return enveloppe

# 2. La fonction classique à décorer
def saluer():
    print("Bonjour de l'entrepôt !")

# 3. Application manuelle du décorateur
# On remplace la fonction d'origine par la version enveloppée
saluer = mon_decorateur(saluer)

# 4. Appel
saluer()

```

---

### 2. **Le Sucre Syntaxique `@`**

L'écriture `saluer = mon_decorateur(saluer)` fonctionne parfaitement, mais elle peut devenir lourde à lire si l'on a beaucoup de fonctions. Python a donc introduit un raccourci visuel (un sucre syntaxique) : le symbole `@`.

Placer `@nom_du_decorateur` juste au-dessus de la définition d'une fonction revient **exactement au même** que l'étape manuelle précédente.

```python
def mon_decorateur(f):
    def enveloppe():
        print("--- Début du bloc décoré ---")
        f()
        print("--- Fin du bloc décoré ---")
    return enveloppe

# Application automatique grâce au sucre syntaxique @
@mon_decorateur
def saluer():
    print("Bonjour via le sucre syntaxique !")

# Appel direct (la fonction est déjà décorée en arrière-plan)
saluer()

```

## 🚀 **LES DÉCORATEURS UNIVERSELS : GESTION DE `*args` ET `**kwargs`**

Jusqu'à présent, nos décorateurs étaient limités : ils enveloppaient uniquement des fonctions simples qui ne prenaient aucun paramètre (comme `saluer()` ou `synchroniser_inventaire()`).

En production, chaque fonction a une signature différente. Si vous tentez d'appliquer un décorateur simple sur une fonction qui prend des arguments, Python lèvera immédiatement une erreur (`TypeError`).

La prochaine étape consiste donc à rendre nos décorateurs **universels** grâce au dépaquetage dynamique.

---

### 1. **Le Problème avec les fonctions paramétrées**

Regardez ce qui se passe si on utilise notre décorateur classique sur une fonction qui prend un argument :

```python
def mon_decorateur(f):
    def enveloppe(): # ❌ N'accepte aucun argument
        print("Opération...")
        f()
    return enveloppe

@mon_decorateur
def calculer_triple(nombre):
    print(nombre * 3)

# 🚨 Lève une TypeError: enveloppe() takes 0 positional arguments but 1 was given
calculer_triple(10) 

```

**Pourquoi l'erreur ?** Parce que le sucre syntaxique `@` a remplacé `calculer_triple` par `enveloppe`. Quand on appelle `calculer_triple(10)`, on appelle en réalité `enveloppe(10)`. Or, `enveloppe` n'est pas taillée pour recevoir des arguments.

---

### 2. **La Solution : L'Enveloppe Universelle**

Pour qu'un décorateur puisse accepter **n'importe quelle fonction**, avec **n'importe quel nombre de paramètres** (positionnels ou nommés), il faut ajouter `*args` et `**kwargs` à deux endroits stratégiques :

1. Dans la signature de la fonction `enveloppe` (pour intercepter tous les arguments envoyés).
2. Lors de l'appel de la fonction d'origine `f(*args, **kwargs)` (pour lui redistribuer fidèlement ses arguments).

Voici le schéma universel :

```python
def decorateur_universel(f):
    # L'enveloppe accepte absolument tout
    def enveloppe(*args, **kwargs):
        print("--- Avant la fonction ---")
        
        # On transmet tout à la fonction d'origine
        resultat = f(*args, **kwargs)
        
        print("--- Après la fonction ---")
        
        # Important : On retourne le résultat généré par la fonction d'origine
        return resultat
        
    return enveloppe

```

---

### 📝 **Cas Pratique : Le Traceur d'Arguments**

Voici une application réelle qui affiche les données reçues par n'importe quelle fonction de notre application :

```python
def espion_de_code(f):
    def enveloppe(*args, **kwargs):
        print(f"🔍 Appel de '{f.__name__}'")
        print(f"   👉 Arguments positionnels : {args}")
        print(f"   👉 Arguments nommés : {kwargs}")
        
        return f(*args, **kwargs)
    return enveloppe

@espion_de_code
def creer_utilisateur(pseudo, role="Membre"):
    print(f"👤 Compte créé pour {pseudo} ({role})")

# Test de l'appel
creer_utilisateur("Baku", role="Admin")

```
